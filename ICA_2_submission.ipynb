```python
# Q.1. Let prices = [250, 99, 475, 120, 310]. Without loops, print the highest price, the lowest price, the total, and the index at which the highest price is stored.
prices = [250, 99, 475, 120, 310]
highest_price = max(prices)
lowest_price = min(prices)
total = sum(prices)
index_of_highest_price = prices.index(highest_price)
print("Highest price:", highest_price)
print("Lowest price:", lowest_price)
print("Total:", total)
print("Index of highest price:", index_of_highest_price + 1)
```

```python
# Q.2. Let a = [1, 2], b = [1, 2] and c = a. Print a == b, a is b and a is c. Add a comment explaining the difference between == and is.
a = [1, 2]
b = [1, 2]
c = a
print(f"{a} == {b}: {a == b}")
print(f"{a} is {b}: {a is b}")
print(f"{a} is {c}: {a is c}")
# "==" shows true if the value in both variables are equal. "is" shows true if and only if is stated that the variables are equal to each other.
```

```python
# Q.3. Loop through the numbers 1 to 10. For even numbers use the pass statement (with a comment saying why pass is used), and for odd numbers print their square.
for i in range(1, 11):
    if i % 2 == 0:
        pass       #pass statement is used to skip the operation to be performed on the value
    else:
        print(i ** 2)
```

```python
# Q.4. Write a function is_vowel(ch) that returns True if the character is a vowel (case-insensitive). Use it to print how many vowels are in the string "Functional Programming".
def is_vowel(ch):
    vowels = "aeiouAEIOU"
    return ch in vowels
string = "Functional Programming"
count = 0
for char in string:
    if is_vowel(char):
        count += 1
print("Number of vowels in the string:", count)
```
