```python
# Question 3
list1 = [1,2,3]
list2 = list1
list2.append(4)
print("list 1:", list1)
print("list 2:", list2)

# When we create list1, we are storing the list in a memory location and the variable 'list1' points to it. 
# When we create list2 and equate it to list1, we are creating two pointers towards the same memory location.
# Hence, when we append list2, we are appending the same list stored in the memory location to which both list1 and list2 are pointing towards.
```
